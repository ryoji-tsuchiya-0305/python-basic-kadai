```python
class Human:
  def __init__(self, name: str, age: int):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人です。")
    else:
      print(f"{self.name}は大人ではありません。")


human_instance = [Human("侍太郎", 30), Human("侍花子", 15), Human("Ryoji", 49), Human("Naoko", 20)]

for human in human_instance:
  human.check_adult()

```
